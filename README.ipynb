```python
m = (b - a) / n
```

```python
aire += f(x_milieu) * m
```

```python
return aire
```

### 1.2 Méthode des rectangles avec NumPy

```python
x = np.linspace(a + m/2, b - m/2, n)
```

## 2. Méthode des trapèzes

### 2.1 Méthode des trapèzes en Python

1. Calculer la largeur d'un segment.

```python
m = (b - a) / n
```

2. Initialiser la somme.

```python
somme = 0
```

3. Parcourir les points intermédiaires.

```python
for i in range(1, n):
```

4. Calculer la position du point.

```python
x = a + i * m
```

5. Ajouter la valeur du polynôme.

```python
somme += f(x)
```

6. Appliquer la formule des trapèzes.

```python
integrale = m * (
    f(a) / 2
    + somme
    + f(b) / 2
)
```

7. Retourner l'intégrale.

```python
return integrale
```

### 2.2 Méthode des trapèzes avec NumPy

1. Générer les points.

```python
x = np.linspace(a, b, n + 1)
```

2. Calculer les valeurs de la fonction.

```python
y = f(x)
```

3. Appliquer la formule vectorisée.

```python
integrale = m * (
    y[0] / 2
    + np.sum(y[1:-1])
    + y[-1] / 2
)
```

4. Retourner le résultat.

```python
return integrale
```
