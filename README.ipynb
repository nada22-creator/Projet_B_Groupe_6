```python
milieu = a + (i + 0.5) * h
```

## 2. Méthode des trapèzes
### 2.1  Méthode des trapèzes en Python

1. Calculer la largeur d'un segment :

   ```python
   m = (b - a) / n
   ```

2. Initialiser la somme à 0.

   ```python
   somme = 0
   ```

3. Parcourir les points intermédiaires avec une boucle.

   ```python
   for i in range(1, n):
   ```

4. Calculer la position du point courant.

   ```python
   x = a + i * m
   ```

5. Évaluer la fonction et ajouter sa valeur à la somme.

   ```python
   somme += f(x, p1, p2, p3, p4)
   ```

6. Appliquer la formule des trapèzes.

   ```python
   integrale = m *  
  
       (f(a, p1, p2, p3, p4) / 2
       + somme
       + f(b, p1, p2, p3, p4) / 2)
   ```

7. Retourner l'intégrale approchée.

   ```python
   return integrale
   ```

## 2.2 Méthode des trapèzes avec NumPy

1. Générer les points d'intégration.

   ```python
   x = np.linspace(a, b, n + 1)
   ```

2. Calculer les valeurs de la fonction.

   ```python
   y = f(x, p1, p2, p3, p4)
   ```

3. Calculer la largeur d'un segment.

   ```python
   m = (b - a) / n
   ```

4. Additionner les valeurs intermédiaires avec NumPy.

   ```python
   np.sum(y[1:-1])
   ```

5. Appliquer la formule vectorisée des trapèzes.

   ```python
   integrale = m * (
       y[0] / 2
       + np.sum(y[1:-1])
       + y[-1] / 2
   )
   ```

6. Retourner l'intégrale approchée.

   ```python
   return integrale
   ```
